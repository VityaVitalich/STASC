```python
from datasets import Dataset, DatasetDict
import json
```

```python
def load_jsonl(path):
    data = []
    with open(path, "r", encoding="utf-8") as f:
        for line in f:
            data.append(json.loads(line))
    return Dataset.from_list(data)

# Load splits
train = load_jsonl("/ltstorage/home/strich/STASC/data/wiki_dev_5000_subsampled.jsonl")
test  = load_jsonl("/ltstorage/home/strich/STASC/data/wiki_test_subsampled.jsonl")

# Rename column if exists
for split in [train, test]:
    if "answers_objects" in split.column_names:
        split = split.rename_column("answers_objects", "references")

# Bundle into DatasetDict
hf_dataset = DatasetDict({
    "train": train,
    "test": test
})

# Inspect

# Push to Hub
hf_dataset.push_to_hub("strich/adaptive_rag_wiki_5000")
```
